```python
import math

# Function to compute binomial probability
def binomial_pmf(n, k, p):
    comb = math.comb(n, k)  # binomial coefficient
    return comb * (p ** k) * ((1 - p) ** (n - k))

# Alice: P(V = 13) where n=52, k=13, p=0.1
alice_prob = binomial_pmf(52, 13, 0.1)

# Bob: P(W = 3) where n=52, k=3, p=0.05
bob_prob = binomial_pmf(52, 3, 0.05)

# Joint probability (assuming independence)
joint_prob = alice_prob * bob_prob

print(f"P(V=13) = {alice_prob:.10f}")
print(f"P(W=3) = {bob_prob:.10f}")
print(f"P(V=13, W=3) = {joint_prob:.10f}")
```
